```python
from pathlib import Path

import pandas as pd
```

```python
#Defines the project folder and raw-data folders for PJM, NYISO, and NOAA.
PROJECT_ROOT = Path.cwd().parent

PJM_DIR = PROJECT_ROOT / "data" / "raw" / "pjm"
NYISO_DIR = PROJECT_ROOT / "data" / "raw" / "nyiso"
NOAA_DIR = PROJECT_ROOT / "data" / "raw" / "noaa"

PJM_PRICE_FILE = PJM_DIR / "da_hrl_lmps_PJM_PS.csv"
PJM_LOAD_FILE = PJM_DIR / "hrl_load_metered_PJM_PS.csv"

NYISO_PRICE_FILE = NYISO_DIR / "nyiso_hudson_valley_jan2025_LMP_DATA.csv"
NYISO_LOAD_FILE = (
    NYISO_DIR
    / "nyiso_hudson_valley_jan2025palIntegrated_HV_loaddata.csv"
)

NEWARK_WEATHER_FILE = NOAA_DIR / "WeatherData Jan25 Newark.csv"
STEWART_WEATHER_FILE = NOAA_DIR / "WeatherData Jan25 Stewart.csv"
```

```python
#Names the six raw CSV files: price and load for each market, plus Newark and Stewart weather.

files = [
    PJM_PRICE_FILE,
    PJM_LOAD_FILE,
    NYISO_PRICE_FILE,
    NYISO_LOAD_FILE,
    NEWARK_WEATHER_FILE,
    STEWART_WEATHER_FILE,
]

# Checks every file exists.
for file in files:
    print(file.exists(), file)
```

```python
#Reads the files into pandas dataframes.
pjm_price_raw = pd.read_csv(PJM_PRICE_FILE)
pjm_load_raw = pd.read_csv(PJM_LOAD_FILE)

nyiso_price_raw = pd.read_csv(NYISO_PRICE_FILE)
nyiso_load_raw = pd.read_csv(NYISO_LOAD_FILE)

# low_memory=False, tells pandas reads large files in chunks before guessing the data types

newark_weather_raw = pd.read_csv(
    NEWARK_WEATHER_FILE,
    low_memory=False,
)

stewart_weather_raw = pd.read_csv(
    STEWART_WEATHER_FILE,
    low_memory=False,
)
```

```python
# creates a dictionary of DataFrames
datasets = {
    "PJM price": pjm_price_raw,
    "PJM load": pjm_load_raw,
    "NYISO price": nyiso_price_raw,
    "NYISO load": nyiso_load_raw,
    "Newark weather": newark_weather_raw,
    "Stewart weather": stewart_weather_raw,
}

#Prints row counts, column counts, and column names.
for name, df in datasets.items():
    print(f"{name:20} rows={df.shape[0]:4} columns={df.shape[1]:3}")
```

```python
# Prints the name of the datasets and fields
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.columns.tolist())
```
